```python
# Import required libraries

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set Seaborn style
sns.set_style("whitegrid")

print("Libraries imported successfully!")
```

```python
# Create Sales Dataset

data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
              "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"],

    "Sales": [12000, 15000, 14000, 18000, 21000, 19000,
              23000, 25000, 22000, 28000, 30000, 32000],

    "Profit": [2500, 3200, 2900, 4000, 4800, 4200,
               5200, 5800, 5000, 6500, 7200, 7800],

    "Orders": [120, 145, 135, 170, 195, 180,
               220, 240, 210, 270, 290, 310]
}

df = pd.DataFrame(data)

# Display the dataset
df
```

```python
# Display first 5 rows

df.head()
```

```python
# Check dataset information

df.info()
```

```python
# Get basic statistical information

df.describe()
```

```python
# Bar Chart - Monthly Sales

plt.figure(figsize=(10, 5))

plt.bar(df["Month"], df["Sales"])

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

```python
# Line Chart - Sales Trend

plt.figure(figsize=(10, 5))

plt.plot(df["Month"], df["Sales"], marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

```python
# Create quarterly sales data

quarter_sales = {
    "Quarter": ["Q1", "Q2", "Q3", "Q4"],
    "Sales": [
        df["Sales"][0:3].sum(),
        df["Sales"][3:6].sum(),
        df["Sales"][6:9].sum(),
        df["Sales"][9:12].sum()
    ]
}

quarter_df = pd.DataFrame(quarter_sales)

quarter_df
```

```python
# Pie Chart - Quarterly Sales

plt.figure(figsize=(7, 7))

plt.pie(
    quarter_df["Sales"],
    labels=quarter_df["Quarter"],
    autopct="%1.1f%%"
)

plt.title("Quarterly Sales Distribution")

plt.show()
```

Insight: Q4 contributes the largest share of annual sales, indicating stronger sales performance toward the end of the year

```python
# Histogram - Sales Distribution

plt.figure(figsize=(8, 5))

plt.hist(df["Sales"], bins=5)

plt.title("Distribution of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")

plt.show()
```

```python
# Scatter Plot - Sales vs Profit

plt.figure(figsize=(8, 5))

sns.scatterplot(
    x=df["Sales"],
    y=df["Profit"]
)

plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")

plt.show()
```

**Key Insights :**

* Sales generally increase throughout the year.   
* December records the highest monthly sales.
* Q4 contributes the largest portion of total annual sales.
* Sales values are mostly concentrated in the higher range.
* Higher sales are generally associated with higher profits.

**Conclusions:**

Data visualization helps us understand large amounts of data quickly. Using bar charts, line charts, pie charts, histograms, and scatter plots, we can identify trends, distributions, comparisons, and relationships in sales data.
