### Observation

The dataset contains student information along with their Mathematics,
Reading, and Writing scores. The numerical columns can be analyzed
using Pandas and NumPy.

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.columns
```

```python
df[['math score', 'reading score', 'writing score']].min()
```

```python
print("Mean Math Score:", np.mean(df['math score']))
print("Mean Reading Score:", np.mean(df['reading score']))
print("Mean Writing Score:", np.mean(df['writing score']))
```

```python
print("Median Math Score:", np.median(df['math score']))
print("Median Reading Score:", np.median(df['reading score']))
print("Median Writing Score:", np.median(df['writing score']))
```

```python
print("Minimum Math Score:", np.min(df['math score']))
print("Maximum Math Score:", np.max(df['math score']))

print("Minimum Reading Score:", np.min(df['reading score']))
print("Maximum Reading Score:", np.max(df['reading score']))

print("Minimum Writing Score:", np.min(df['writing score']))
print("Maximum Writing Score:", np.max(df['writing score']))
```

```python
print("Math Standard Deviation:", np.std(df['math score']))
print("Reading Standard Deviation:", np.std(df['reading score']))
print("Writing Standard Deviation:", np.std(df['writing score']))
```

```python
print(df['math score'].mean())
print(df['reading score'].mean())
print(df['writing score'].mean())
```

```python
print("Highest Math Score:", df['math score'].max())
print("Highest Reading Score:", df['reading score'].max())
print("Highest Writing Score:", df['writing score'].max())
```

```python
print("Lowest Math Score:", df['math score'].min())
print("Lowest Reading Score:", df['reading score'].min())
print("Lowest Writing Score:", df['writing score'].min())
```

```python
df[df['math score'] > 80].head()
```

```python
df.sort_values('math score', ascending=False).head()
```
