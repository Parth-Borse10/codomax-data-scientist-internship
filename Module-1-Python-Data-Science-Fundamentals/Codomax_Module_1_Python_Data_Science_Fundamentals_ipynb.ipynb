## 1. Variables and Data Types

Variables are used to store data in Python. Python supports different data types such as integers, floats, strings, and boolean values.

```python
# Basic variables and data types

name = "Parth"
age = 21
percentage = 87.5
is_student = True

print("Name:", name)
print("Age:", age)
print("Percentage:", percentage)
print("Student:", is_student)

print("\nData Types:")
print("name:", type(name))
print("age:", type(age))
print("percentage:", type(percentage))
print("is_student:", type(is_student))
```

## 2. Lists and Tuples

Lists and tuples are used to store multiple values. Lists are mutable, while tuples are immutable.

```python
# Working with a list

marks = [78, 85, 92, 88, 76]

print("Marks:", marks)
print("First mark:", marks[0])
print("Total marks:", sum(marks))
print("Average marks:", sum(marks) / len(marks))
```

```python
# Working with a tuple

subjects = ("Python", "Statistics", "Machine Learning", "SQL")

print("Subjects:", subjects)
print("First subject:", subjects[0])
print("Number of subjects:", len(subjects))
```

## 3. Dictionaries

A dictionary stores data as key-value pairs and is useful for representing structured information.

```python
# Working with a dictionary

student = {
    "name": "Parth",
    "age": 21,
    "course": "AI & Data Science",
    "score": 87
}

print("Student details:")

for key, value in student.items():
    print(f"{key}: {value}")
```

## 4. Conditional Statements

Conditional statements allow a program to make decisions based on conditions.

```python
# Using if-elif-else

score = 87

if score >= 90:
    grade = "A+"
elif score >= 75:
    grade = "A"
elif score >= 60:
    grade = "B"
else:
    grade = "C"

print("Score:", score)
print("Grade:", grade)
```

## 5. Loops

Loops are used to repeatedly execute a block of code.

```python
# For loop

print("For loop:")

for number in range(1, 6):
    print(number)
```

```python
# While loop

print("While loop:")

count = 1

while count <= 5:
    print(count)
    count += 1
```

## 6. Functions

Functions are reusable blocks of code that perform a specific task.

```python
# Creating and using a function

def calculate_average(values):
    """Calculate the average of numeric values."""

    if len(values) == 0:
        return 0

    return sum(values) / len(values)


numbers = [10, 20, 30, 40, 50]

print("Numbers:", numbers)
print("Average:", calculate_average(numbers))
```

## 7. File Handling

File handling allows Python programs to create, write, read, and process files.

```python
# Writing data to a text file

file_name = "student_data.txt"

with open(file_name, "w") as file:
    file.write("Name,Score\n")
    file.write("Parth,87\n")
    file.write("Rahul,82\n")
    file.write("Aman,91\n")

print("File created successfully.")
```

```python
# Reading data from the file

with open("student_data.txt", "r") as file:
    content = file.read()

print(content)
```

## 8. NumPy Basics

NumPy is a Python library used for numerical computing and efficient array operations.

```python
# Basic NumPy operations

import numpy as np

values = np.array([10, 20, 30, 40, 50])

print("Array:", values)
print("Mean:", np.mean(values))
print("Minimum:", np.min(values))
print("Maximum:", np.max(values))
print("Standard Deviation:", np.std(values))
```

## 9. Pandas Basics

Pandas is widely used in Data Science for data manipulation, analysis, and working with tabular datasets.

```python
# Create a simple DataFrame

import pandas as pd

data = {
    "Name": ["A", "B", "C", "D"],
    "Age": [20, 22, 21, 23],
    "Score": [78, 91, 85, 88]
}

df = pd.DataFrame(data)

df
```

```python
# Basic Pandas analysis

print("First rows:")
display(df.head())

print("\nDataset information:")
df.info()

print("\nAverage score:", df["Score"].mean())
print("Highest score:", df["Score"].max())
```
